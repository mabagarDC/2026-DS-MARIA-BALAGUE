## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def DiasSemana (numero:int):

    if numero == 1:
        dia = "Lunes"
    elif numero == 2:
        dia = "Martes"
    elif numero == 3:
        dia = "Miercoles"
    elif numero == 4:
        dia = "Jueves"
    elif numero == 5:
        dia = "Viernes"
    elif numero == 6:
        dia = "Sabado"
    elif numero == 7:
        dia = "Domingo"        
    else:
        dia = "Número no válido"
    return dia 

quedia = DiasSemana(10)

print(quedia)
````

## Ejercicio 2
En el ejercicio 8 de bucles, creábamos una pirámide invertida, cuyo número de pisos venía determinado por un input del usuario. Crear una función que replique el comportamiento de la pirámide, y utiliza un único parámetro de entrada de la función para determinar el número de filas de la pirámide, es decir, elimina la sentencia input.

````python

def piramide (numero:int):
    ejer_8 = list(range(numero))

    #tendremos que hacer un bucle, pero que empiece en la posicion n 
    # Sumamos 1 a cada elemento porque range(numero) empieza en 0
    for i in range(len(ejer_8)):
        ejer_8[i] = ejer_8[i] + 1

    # Imprimimos la lista y vamos eliminando el último elemento
    while len(ejer_8) > 0:
        print(*ejer_8)
        ejer_8.pop()

piramide(6)
````

## Ejercicio 3
Escibe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero se  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def comparacion(arg1:int, arg2:int): 
   
    if arg1 == arg2:
       return print("numeros iguales")

    elif arg1 > arg2:
       return print ("El primer numero es mayor que el segundo")

    return print("el segundo es mayor que el primero")


comparacion(8,9)
comparacion(10,9)
comparacion(8,8)
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea la letra a contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayuscula, como minúscula

````python

#Indicar la letra en minuscula
def contador_letras (arg1:str, arg2:str):

    #Hacer de la frase una lista 
    frase =list(arg1)
 

    #Inicializar el contador a 0
    contador = 0

    #Recorrer la lista 
    for letra in frase:

        #donde compararemos la letra que nos indica tanto en minuscula como en mayuscula 
        if letra == arg2 or letra == arg2.upper():

            #en caso de que si se encuentre se suma el contador 
            contador = contador + 1 
            
    #devuelve el numero de letra que ha contado 
    return contador

contador_letras("En un lugar de La Mancha, de cuyo nombre no quiero acordarme", "a")
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string.

````python
def cuenta_letras (arg1:str):
    
      #Hacer de la frase una lista 
      lista_letras = list(arg1)
      
      #Inicialización del diccionario, donde se va a guardar clave(letra) y valor (contador)
      diccionario = {}

      #Recorrer la lista 
      for letra in lista_letras: 
            
            #Si la letra todavía no existe en el diccionario, la añadimos con valor 1
            if letra in diccionario:
                  diccionario[letra] = diccionario[letra]+1

            #Si la letra todavía no existe, la añadimos con valor 1
            else:
                 diccionario[letra] = 1

      #devuelve el diccionario
      return diccionario


cuenta_letras("MARIAAMM")
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

````python
#arg1 es una lista, arg2 se debe de indicar el comando add o remove, y el arg3 es por defecto none
def modificar_listas (arg1:list, arg2:str, arg3:None):
    
    #Primero creamos una copia del original para no modificar la entrada 
    auxiliar = arg1.copy()
    
    #Primero se comprueba si el argumento 2 que tipo de comando ha indicado 
    if arg2== "add":
        
        #Si es add entonces append añade el elemento arg3
        auxiliar.append(arg3)
    
    #en cambio si indica remove (arg2)
    elif arg2== "remove" :
        auxiliar.remove(arg3)

    #Y en caso que en el arg2 no indique ni add ni remove que ponga en pantalla que se ha equivocado
    else:
        print("El comando debe ser 'add' o 'remove'")
    
    return auxiliar


modificar_listas ([1, 2, 5], "add", 3)
````

## Ejercicio 7
Crea una función que reciba un número arbitrario de palabras, y devuelva una frase completa, separando las palabras con espacios.

````python
# Esta funcion le puedes pasar un número variable de strings
def palabras (*args:str): 

    #Inicializamos un string vacío donde iremos construyendo la frase
    string_palabra = ""

    #Recorremos los argumentos obteniendo el índice y la palabra
    for i,palabra in enumerate(args):

        #Si es la primera palabra
        if i ==0 :
            
            #La añadimos sin espacio delante
            string_palabra += palabra 

        #Si no es la primera palabra
        else:
            #Añadimos un espacio delante y después la palabra
            string_palabra += " " + palabra 
    
    #Devolvemos la frase completa
    return string_palabra

palabras("me", "llamo", "Maria")

````
