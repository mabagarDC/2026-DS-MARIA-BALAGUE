## Ejercicio 5
Realiza un conversor de grados Celsius a Fahrenheit. El conversor tiene que recibir el input del usuario y devolver la conversión.

La conversión de grados sigue la siguiente fórmula: F = (C × 9 / 5) + 32 

```python
c = input ("Me puedes indicar que temperatura hace en Celsius(decimales en formato punto)?: ")

#Primero tranformado el string a numero decimal
cdecimal = float(c)
#Calcula la conversión a Fahrenheit
F = (cdecimal*9/5)+32
#Sale por pantalla el resultado final, con f y los corchetes enlazamos con la variable deseada
print (f"El equivalente de grados a Fahrenheit es {F}")
```
