## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

````python
grados = input ("Hola, ¿Me puedes indicar que grados quieres transformar a radianes(decimales en formato punto)?: ")
#importamos el modulo math
import math
#Tranformamos la variable grados de string a float
grados = float(grados)
#Calcula los radianes
radianes = round(grados*(math.pi/180), 2)
#Sale en pantalla el resultado
print (f"Corresponde a {radianes} radianes")
````

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

````python
base = float(input("Hola, para calcular el area de un paralelogramos necesitamos saber la longitud de la base, ¿me la puedes indicar en metros? "))
altura = float(input ("ahora indicame la altura en metros: "))
#Calcula el area y redondea con dos decimales el resultado final
area= round(base*altura, 2)
print (f"El area total es {area} m^2")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
frase = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
print(frase)
print("Todo mayusculas:", frase.upper())
print("Todo minusculas:", frase.lower())
print("Sustituir las primeras letras de minuscula a mayuscula:", frase.replace("En un lugar de la Mancha, de cuyo nombre no quiero acordarme.", "En Un Lugar De La Mancha, De Cuyo Nombre No Quiero Acordarme."))
print("Separalo segun el numero de espacios:", frase.split(" "))
print("Sustituir , por ;:", frase.replace(",", ";"))
print("Eliminar a:", frase.replace("a", ""))
````
