## Ejercicio 3
1. Escribe un programa que vaya pidiendo numeros al usuario. Cuando el usuario introduzca el 0, el programa tiene que imprimir por pantalla el sumatorio de todos los numeros positivos introducidos
2. Además de la suma, queremos que imprima también una lista con todos los números.

```python
numero = int(input("Dime un número: "))
suma = 0
numeros = []

while numero !=0:

    if numero > 0:
        suma = suma + numero
        
    numeros.append(numero)
    numero = int(input("Dime otro número: "))

print("La suma es:", suma)
print("Los numeros son:" , numeros)
```

## Ejercicio 5
Imprime por pantalla la palabra "Python", pero empezando con la última letra.

```python
#Pasamos la palabra en lista 
palabra = list("Python")

#Creamos una lista nueva
reves =[]

#tendremos que hacer un bucle, pero que empiece en la posicion n 
for n in range(len(palabra)-1, -1,-1):

#Vamos añadiendo la palabra al reves 
   reves.append(palabra[n]) 

#Se vuelve a juntar cada letra en un solo string
"".join(reves)
    
```

## Ejercicio 6
Escribe un programa que calcule la cantidad de veces que está la letra "m" o "M" en la frase:
> "En un lugar de La Mancha, de cuyo nombre no quiero acordarme"

```python
frase =list("En un lugar de La Mancha, de cuyo nombre no quiero acordarme")
contador = 0

for letra in frase:

    if letra == "M" or letra == "m":
        
        contador = contador + 1 

print (contador)
```
