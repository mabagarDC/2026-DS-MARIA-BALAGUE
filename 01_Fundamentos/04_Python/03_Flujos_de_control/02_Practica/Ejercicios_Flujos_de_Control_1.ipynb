## Ejercicio 1
Declara una variable numérica que será una hora cualquiera del día. Implementa mediante sentencias `if/elif/else` la siguiente casuística:

* Si es entre las 0 y las 8, print "Durmiendo"
* Si es entre las 9 y las 18, print "Trabajando"
* Si es entre las 19 y las 21, print "Clase"
* Si es entre las 22 y las 24, print "Descanso"
* En cualquier otro caso, print "Transporte o error"

```python

#EJERCICIO 1

HoraUsuario = int(input("Hola, me podría indicar la hora del día?: "))

#Primero comprobamos que no nos indiquen numeros raros 
if HoraUsuario <0 or HoraUsuario > 24:
     print("Error")
#Compromacones de las hora proporcionada
elif  0 <= HoraUsuario<= 8:
    print("Durmiendo")

elif  9 <= HoraUsuario<= 18:
    print("Trabajando")

elif  19 <= HoraUsuario<= 21:
    print("Clase")
    
elif  22 <= HoraUsuario<= 24:
    print("Descanso")
    
```

## Ejercicio 2
En este ejercicio vamos a implementar un calculador de precios de casas muy sencillo. Tenemos las siguientes variables:

> superficie
>
> distrito

Implementa mediante sentencias `if/elif/else` la siguiente casuística:

1. Si el distrito es "Moncloa" o "Centro", y además la superficie es superior a 100 metros cuadrados, el precio de la casa es de 1000
2. Si el distrito es "Salamanca", y además la superficie de la casa es al menos de 150 metros, el precio de la casa es de 1500
3. Si el distrito no es "Retiro" y la superficie está entre 60 y 80 metros, el precio es de 600
4. En cualquier otro caso, el precio será de 0

```python
#EJERCICIO 2
superficie = int(input("Hola, indicame que superficie estaría interesado para un piso: "))
distrito = input("Ahora indicame que distrito le gustaría que estuviera el piso: ")

distrito_opciones = ["Moncloa", "Centro" ,"Centro", "Salamanca", "Retiro"]

#Una primera condicion donde, se comprueba que las opciones que nos proporciona el input sea dentro de nuestras opciones 
if (superficie<60 or 80<superficie<100) or distrito not in distrito_opciones:
    print("Precio = 0 €")
#Empezamos las condiciones establecidas del ejercicio
elif superficie >= 150 and distrito == "Salamanca":
    print("Precio = 1500 €")
elif superficie >=100 and (distrito == "Moncloa" or distrito == "Centro"):
    print("Precio = 1000 €")
elif 60<=superficie<=80 and distrito == "Retiro" :
    print("Precio = 600 €")
```

## Ejercicio 4
Escribe un programa que tenga dos variables: un numero, y una lista numérica. El programa debe recorrer la lista e imprimir por pantalla cada elemento de la lista multiplicado por el número

```python

num = int(input("Hola, dime un número: "))

numeros = []

# Para guardar una lista de cinco números,
# creo un bucle que pide 5 números al usuario
for i in range (5):
    numero =int(input("Introduce un número: "))
    numeros.append(numero)
#Recorro la lista y multiplico cada elemento por num
for numero in numeros:
    print(numero*num)

```

## Ejercicio 6
Dada la siguiente lista:

> list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

Implementa un programa que los recorra e imprima por pantalla todos los divisibles por 5. Si nos encontramos con alguno que sea mayor que 150, detener el bucle.

```python
list1 = [12, 15, 32, 42, 55, 75, 122, 32, 150, 180, 200]
#este bucle mira cada valor de la lista 
for n in list1: 
    if n > 150 :
        break
    if n % 5 == 0:
        print(n)
```
