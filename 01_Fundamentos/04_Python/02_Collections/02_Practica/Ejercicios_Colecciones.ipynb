## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output:

['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]

#Primero se tiene que inicializar la nueva lista
resultado1 = []
resultado2 = []

# Creo dos listas para guardar por separado las combinaciones
# palabra1 + palabra2 y palabra2 + palabra1
for palabra1 in ejer_3_1:
    for palabra2 in ejer_3_2:
        resultado1.append(palabra1+ " "+ palabra2 )
        resultado2.append(palabra2+ " "+ palabra1 )

resultadoF = resultado1 + resultado2

print(resultadoF)
````

## Ejercicio 4
Dada la siguiente lista, encuentra el índice del valor 45 y sustitúyelo por el 0
> ```ejer_4 = [20, 47, 19, 29, 45, 67, 78, 90]```

````python
ejer_4 = [20, 47, 19, 29, 45, 67, 78, 90]

#La manera fácil sin bucle
#ejer_4[4] = 0
#print(ejer_4)

#Para poder activar la menra del bucle, se debe de comentar el anterior.
#Para recorrer la lista teniendo en cuenta la posición, o dicho de otra manera, si tienes que mirar que valor tiene cada índice,
#tienes que indicarlo: range(len(lista))
#FORMA COMPLICADA Y MUY POCO EFICIENTE
for i in range(len(ejer_4)):
    if ejer_4[i] == 45:
        ejer_4[i] = 0

#FORMA FACIL 


print(ejer_4)
    
````

## Ejercicio 12
Convierte la siguiente tupla en un diccionario
> ```ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("y", 6), ("z", 1)]

#Crear un diccionario vario
diccionario = {}

#Recorro la tupla, donde el primer elemento es la clave y el segundo el valor
#
for clave, valor in ejer_12:
#Si la clave no existe en el diccionario, crea una lista de valores
    if clave not in diccionario:
        diccionario[clave] = []
#En caso de que si exista, se añadirá a la lista de esa clave otro valor
    diccionario[clave].append(valor)

print(diccionario)


````
