# Name of the max valued product

```python
# 1. PRODUCTO CON MAYOR PRECIO UNITARIO

# Calculamos el precio de una unidad dividiendo
# el precio total de la línea entre la cantidad comprada
chipo["unit_price"] = chipo["item_price"] / chipo["quantity"]

# Buscamos el índice del producto con mayor precio unitario
# y mostramos su nombre
chipo.loc[chipo["unit_price"].idxmax(), "item_name"]
```

```python
#de manera comprimida

chipo.loc[chipo["item_price"].idxmax(), "item_name"]
```
